# **3. Apresentação aos Transformadores**

Descrito pela primeira vez em 2017 no trabalho [*Attention is all you need*](https://arxiv.org/pdf/1706.03762), o modelo de rede neural utilizando Transformers revolucionou o NLP (Natural Language Processing, ou Processamento de Linguagem Natural), que consiste na forma como inputs textuais são interpretados pela máquina. Ao contrário dos modelos utilizados anteriormente, como as redes neurais recorrentes (RNN's) e as redes neurais convolucionais (CNN's), os transformadores são capazes de processar sequências de input em paralelo, tornando-os altamente eficientes para treinamento e inferência. Suas principais inovações são:

* **Codificação Posicional**: em vez de "olhar" para cada palavra na ordem em que aparece na frase, um número exclusivo é atribuído a cada palavra, de forma a fornecer informações sobre a posição de cada token na sequência, permitindo uma interpretação sequencial da sequência de palavras.
* **Autoatenção**: a atenção, conforme apresentado pelo [mecanismo de Bahdanau](https://arxiv.org/pdf/1409.0473), calcula pesos para cada palavra em uma frase conforme elas se relacionam com as outras palavras na frase. a fim de prever as que serão usadas na sequência. A autoatenção, por sua vez, faz com que cada palavra considere todas as outras na sequência simultaneamente, avaliando sua importância para o token atual.

A imagem apresentada abaixo, trazida por *Vaswani et al.*, resume a arquitetura do modelo de Transformadores.

![](https://d1.awsstatic.com/GENAI-1.151ded5440b4c997bac0642ec669a00acff2cca1.png)

Nesse diagrama, os principais blocos a serem observados, conforme apresentado pela AWS na matéria [*O que são transformadores em inteligência artificial?*](https://aws.amazon.com/pt/what-is/transformers-in-artificial-intelligence/), são descritos resumidamente a seguir:

* **Incorporações de Entrada:** a sequência de entrada é convertida em linguagem matemática, sendo dividida em tokens para que eles sejam transformados em vetores. Estes, carregam informações semânticas e de sintaxe, representados por números para serem aprendidos no treinamento.
* **Codificação Posicional:** a arquitetura do Transformador, por não processar os dados de entrada de uma forma sequencial, utiliza a Codificação Posicional para considerar a ordem dos tokens. Nela, o modelo gera um sinal posicional exclusivo que é adicionado ao vetor durante o processo de Incorporação, junto às outras informações do token.
* **Bloco de Transformador:** responsável pela avaliação da importância dos termos em uma frase, a fim de treinar o modelo. É composto de dois mecanismos principais: o de autoatenção "com várias cabeças", que fornece a relevância de cada símbolo na sequência, e uma rede de feedback para a posição de cada elemento.
* **Blocos Lineares:** executa um mapeamento linear do espaço vetorial para o domínio de entrada, isto é, transforma as representações internas em previsões concretas para gerar um conjunto de pontuações para cada token.
* **Softmax:** etapa final, na qual as pontuações calculadas são covertidas em uma distribuição normalizada, de forma a indicar a confiança do modelo em uma classe ou token.

Vamos aplicar um filtro para remover elementos sem nota ou sem texto, além de eliminar as demais informações do dataset:

## 4.3. Pré-Processamento de Dados e Tokenização

Aqui, há um processamento prévio dos dados com o fim de padronizar as entradas e auxiliar o aprendizado.

Primeiramente, tokenizamos o texto e construímos um vocabulário baseado na frequência dos tokens:

```python
import re # Biblioteca para definição de padrões de busca e extração no texto
from collections import Counter # Biblioteca para contagem de tokens

# Função de tokenização básica (usando regex para extrair tokens alfanuméricos em lowercase)
def tokenize_text(text):
    tokens = re.findall(r'\w+', text.lower())
    return tokens

# Construção do vocabulário utilizando o dataset de treino
all_tokens = []
for exemplo in dataset_train['text']:
    all_tokens.extend(tokenize_text(exemplo))

vocab_counter = Counter(all_tokens)
vocab = [token for token, count in vocab_counter.items() if count >= 1]

# Reservando índices especiais: 0 para <PAD> e 1 para <UNK>
token2idx = {token: idx+2 for idx, token in enumerate(vocab)}
token2idx["<PAD>"] = 0
token2idx["<UNK>"] = 1

print(f"Tamanho do vocabulário: {len(token2idx)} tokens.")
```

Depois, definimos as funções para conversão dos textos em sequências numéricas, com truncamento e padding.

Além disso, criamos também a criação da Máscara de Atenção, responsável por indicar a relevância de cada posição:

```python
# Função para codificar um texto, aplicando truncamento e padding
def encode_text(text, token2idx, max_length=256):
    tokens = tokenize_text(text)
    encoded = [token2idx.get(token, token2idx["<UNK>"]) for token in tokens]

    if len(encoded) > max_length:
        encoded = encoded[:max_length]  # Trunca, se necessário
    else:
        encoded += [token2idx["<PAD>"]] * (max_length - len(encoded))  # Adiciona padding, se necessário
    return encoded

# Função para criar a máscara de atenção (1 para token real e 0 para padding)
def create_attention_mask(encoded_text):
    return [1 if token != token2idx["<PAD>"] else 0 for token in encoded_text]

# Função para aplicar a codificação em cada exemplo do dataset
def encode_example(example):
    encoded = encode_text(example["text"], token2idx, max_length=256)
    example["input_ids"] = encoded
    example["attention_mask"] = create_attention_mask(encoded)
    return example
```

Por fim, aplicamos os tratamentos nos conjuntos de treino e de teste:

```python
# Aplica a tokenização customizada nos datasets de treino e teste
dataset_train = dataset_train.map(encode_example)
dataset_test = dataset_test.map(encode_example)

print("Exemplo tokenizado:", dataset_train[1])  # Exemplo
```

## 4.4. Construção do Transformador

Aqui, vamos iniciar a concepção do modelo de Transformador, seguindo seus principais elementos. Atente-se para os componentes descritos em [*3. Apresentação aos Transformadores*](), e como eles aparecerão nessa etapa:

```python
import torch  # Biblioteca do PyTorch para tensores e treinamento de modelos
import torch.nn as nn  # Fornece componentes para a construção de redes neurais (camadas, funções de ativação, etc.)
from torch.utils.data import DataLoader  # Facilita a criação de batches
import math  # Biblioteca para cálculos na etapa de codificação posicional
import numpy as np  # Biblioteca para operações numéricas e manipulação de arrays
```

Após as importações, iniciamos a construção do Transformador com a criação da classe que atuará como o Codificador Posicional:

```python
# Classe que atuará como Codificador Posicional
class PositionalEncoding(nn.Module):

    def __init__(self, d_model, dropout=0.1, max_len=5000):
        super().__init__()
        self.dropout = nn.Dropout(p=dropout)

        # Cria uma matriz [max_len, d_model] para armazenar as codificações posicionais
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)

        # Calcula o termo de escala para as funções seno e cosseno
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))

        # Aplica as funções em cada posição
        pe[:, 0::2] = torch.sin(position * div_term)  # Seno para índices pares
        pe[:, 1::2] = torch.cos(position * div_term)  # Cosseno para índices ímpares

        # Atribui a forma [1, max_len, d_model]
        pe = pe.unsqueeze(0)

        # Registra o tensor como um parâmetro não treinável
        self.register_buffer('pe', pe)

    def forward(self, x):  # O parâmetro é um tensor de forma [batch_size, seq_len, d_model]

        # Adiciona as codificações de cada posição (e corta, se necessário)
        x = x + self.pe[:, :x.size(1)]

        return self.dropout(x)
```

Agora, vamos à criação da classe que atuará como o bloco de Transformador:

```python
# Classe que atuará como bloco de Transformador
class TransformerRegressor(nn.Module):

    def __init__(self, vocab_size, d_model=256, nhead=16, num_encoder_layers=4, dim_feedforward=256, dropout=0.1, max_seq_length=256):
        super().__init__()

        # Camada de incorporação, que converte os índices em vetores densos
        self.embedding = nn.Embedding(vocab_size, d_model)

        # Adiciona as codificações posicionais
        self.pos_encoder = PositionalEncoding(d_model, dropout, max_seq_length)

        # Camada encoder do Transformador (com autoatenção)
        encoder_layer = nn.TransformerEncoderLayer(d_model, nhead, dim_feedforward, dropout, batch_first=True)
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_encoder_layers)

        # Camada de normalização
        self.norm = nn.LayerNorm(d_model)

        # "Head de regressão": camada final, reduz a representação para um único valor
        self.regressor = nn.Linear(d_model, 1)
        self.d_model = d_model

    def forward(self, src, src_key_padding_mask=None):  # Os parâmetros são tensores de forma [batch_size, seq_length]

        # Incorporação e escala
        src = self.embedding(src) * math.sqrt(self.d_model)

        # Adiciona codificação posicional
        src = self.pos_encoder(src)

        # Transpõe o tensor para o formato esperado: [seq_length, batch_size, d_model]
        src = src.transpose(0, 1)

        # Aplica o encoder do Transformer
        output = self.transformer_encoder(src, src_key_padding_mask=src_key_padding_mask)

        # Seleciona a representação do primeiro token
        output = output.mean(dim=0)  # Média dos embeddings da sequência

        # Normaliza a saída antes de alimentá-la à regressão
        output = self.norm(output)

        # Passa pela camada linear para gerar a predição contínua
        output = self.regressor(output)  # [batch_size, 1]

        return output
```

Aqui, é criada uma máscara que será utilizada pelo modelo para ignorar os tokens de padding durante o aprendizado:

```python
# Cria uma máscara para indicar as posições de padding
def create_src_key_padding_mask(input_ids):  # O parâmetro é um tensor de forma [batch_size, seq_len]

    return (input_ids == 0)
```

Antes da criação do modelo em si, é importante verificar em qual tipo de dispositivo o ambiente de execução está rodando:

```python
# Verifica qual dispositivo está sendo usado pelo ambiente
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Usando dispositivo: {device}")
```

Agora, seguimos para a criação do modelo utilizando os parâmetros definidos pelo usuário:

```python
vocab_size = len(token2idx)  # O tamanho do vocabulário já foi definido anteriormente

# Criação do modelo, seguindo os parâmetros definidos no tópico 4.1.
model = TransformerRegressor(
    vocab_size=vocab_size,
    d_model=d_model,
    nhead=nhead,
    num_encoder_layers=num_encoder_layers,
    dim_feedforward=dim_feedforward,
    dropout=dropout,
    max_seq_length=max_seq_length
).to(device)  # Migra o modelo para o dispositivo do ambiente
```

## 4.5. Etapa de Treinamento

Nessa etapa, vamos direcionar a nossa base de dados já processada para o Transformador, a fim de treinar o modelo.

Primeiramente, definimos uma função para a conversão de uma lista de exemplos em um batch, a fim de padronizar a entrada do modelo:

```python
# Supondo que dataset_train já possua 'input_ids' e 'attention_mask'
def collate_fn(batch):

    # Converte as listas para tensores de forma [batch_size, seq_length]
    input_ids = torch.tensor([example["input_ids"] for example in batch])

    # Cria a máscara de padding
    src_key_padding_mask = (input_ids == token2idx["<PAD>"])

    # Garante que a máscara seja do tipo booleano e a move para o dispositivo correto
    src_key_padding_mask = src_key_padding_mask.t().to(torch.bool).to(device)

    # Extrai as notas de cada exemplo, converte para float e adiciona uma dimensão extra
    ratings = torch.tensor([example["rating"] for example in batch], dtype=torch.float).unsqueeze(1)

    return input_ids, src_key_padding_mask, ratings
```

Depois, colocamos as definições para o otimizador (Adam) e para a função de perda (MSE):

```python
optimizer = torch.optim.Adam(model.parameters(), lr=3e-5)  # Define o otimizador (Adam)
criterion = nn.MSELoss()  # Define a função de perda (MSE)
```

```python
# Cria um DataLoader para o conjunto de treino
train_loader = DataLoader(dataset_train, batch_size=batch_size, shuffle=True, collate_fn=collate_fn)

# Coloca o modelo em modo de treino
model.train()
```

É importante migrar, se necessário, os tensores para a GPU, com a finalidade de acelerar o treinamento:

```python
for batch in train_loader:
    input_ids, src_key_padding_mask, ratings = [x.to(device) for x in batch]  # Move os batches para a GPU
    # ratings = (ratings - 1) / 4
    optimizer.zero_grad()
    outputs = model(input_ids, src_key_padding_mask=src_key_padding_mask)
    loss = criterion(outputs, ratings)
    loss.backward()
    optimizer.step()
```

Em seguida, nos direcionamos para o treinamento em si:

```python
# Execução do treinamento
for epoch in range(epochs):

    # Variáveis de controle do treinamento
    epoch_loss = 0.0
    epoch_acc = 0.0
    num_samples = 0

    for batch in train_loader:
        input_ids, src_key_padding_mask, ratings = [x.to(device) for x in batch]  # Obtém o batch: input_ids, máscara de padding e os ratings reais
        optimizer.zero_grad()  # Zera os gradientes do otimizador
        # ratings = (ratings - 1) / 4  # Normaliza os ratings para [0,1]
        outputs = model(input_ids, src_key_padding_mask=src_key_padding_mask)  # Gera a predição a partir do batch de input_ids (Forward Pass)
        loss = criterion(outputs, ratings)  # Calcula a perda
        loss.backward()  # Computa os gradientes (Backward Pass)
        optimizer.step()  # Atualiza os pesos do modelo

        # Acumula a perda total ponderada pelo número de exemplos do batch
        batch_size = input_ids.size(0)
        epoch_loss += loss.item() * batch_size

        # Calcula a "acurácia" do batch, considerando "correta" a previsão quando a diferença entre o valor previsto e o real for menor que a tolerância
        batch_correct = (torch.abs(outputs - ratings) < tolerance).float().sum().item()
        epoch_acc += batch_correct
        num_samples += batch_size

    # Calcula a perda média e a "acurácia" média para a época
    epoch_loss /= num_samples
    epoch_acc /= num_samples

    print(f"Epoch {epoch+1} - Loss: {epoch_loss:.4f} - Accuracy: {epoch_acc*100:.2f}%")
```

## 4.6. Etapa de Testes

Nesta etapa, após o treinamento, vamos utilizar o modelo já treinado para fazer previsões usando o conjunto de teste.

```python
# Cria um DataLoader para o conjunto de teste
test_loader = DataLoader(dataset_test, batch_size=batch_size, shuffle=False, collate_fn=collate_fn)

# Coloca o modelo em modo de avaliação
model.train()
```

Assim como na etapa de treinamento, deve-se mover os batches para a GPU:

```python
with torch.no_grad():
    for batch in test_loader:
        input_ids, src_key_padding_mask, ratings = [x.to(device) for x in batch]  # Move os batches para a GPU
        outputs = model(input_ids, src_key_padding_mask=src_key_padding_mask)
```

```python
# Cria listas para comparação futura
all_actuals = []
all_predictions = []

# Variáveis para o acúmulo de métricas
test_loss = 0.0
test_acc = 0.0
num_samples = 0

# Desativa o cálculo de gradientes para acelerar a avaliação e economizar memória
with torch.no_grad():
    for batch in test_loader:

        # Obtém o batch: input_ids, máscara de padding e os ratings reais
        input_ids, src_key_padding_mask, ratings = [x.to(device) for x in batch]

        # Faz a previsão com o modelo
        outputs = model(input_ids, src_key_padding_mask=src_key_padding_mask)

        # Converte outputs de volta para a escala [1,5]
        # outputs = outputs * 4 + 1  # Desnormaliza as previsões

        # Calcula a perda do batch
        loss = criterion(outputs, ratings)

        # Acumula a perda ponderada pelo tamanho do batch
        batch_size = input_ids.size(0)
        test_loss += loss.item() * batch_size

        # Calcula a "acurácia" do batch: fração dos exemplos cuja previsão está dentro da tolerância
        batch_acc = (torch.abs(outputs - ratings) < 0.5).float().mean().item()
        test_acc += batch_acc * batch_size
        num_samples += batch_size

        # Armazena os valores reais e previstos para análise posterior
        all_predictions.extend(outputs.squeeze().tolist())
        all_actuals.extend(ratings.squeeze().tolist())

# Converte as listas para numpy arrays
actuals = np.array(all_actuals)
predictions = np.array(all_predictions)

# Calcula as métricas médias no conjunto de teste
test_loss /= num_samples
test_acc /= num_samples

print(f"Test Loss: {test_loss:.4f} - Test Accuracy: {test_acc*100:.2f}%")
```

## 5.1. Matriz de Confusão

Para fins de visualização do desempenho do modelo de forma prática, vamos gerar uma Matriz de Confusão para comparar os resultados esperados com os resultados reais presentes no conjunto de testes, utilizando a tolerância definida anteriormente:

```python
import matplotlib.pyplot as plt  # Biblioteca para plotagem de gráficos
import seaborn as sns  # Biblioteca para visualização de dados com gráficos
```

```python
print(f'Tolerância: {tolerance}\n') # Ressaltando o valor definido para tolerância

bins = np.arange(1, 5 + tolerance, tolerance)  # Gera os intervalos

# Calcula a matriz de contagem 2D que agrupa as ocorrências para cada par (valor real, valor previsto)
heatmap_matrix, xedges, yedges = np.histogram2d(actuals, predictions, bins=[bins, bins])

# Calcula os valores centrais dos bins para os rótulos dos eixos
x_labels = (xedges[:-1] + xedges[1:]) / 2
y_labels = (yedges[:-1] + yedges[1:]) / 2

# Plota o heatmap:
plt.figure(figsize=(8, 6))
sns.heatmap(heatmap_matrix, annot=True, fmt='.0f', cmap='Blues', xticklabels=np.round(x_labels, 2), yticklabels=np.round(y_labels, 2))
plt.xlabel('Valor Previsto')
plt.ylabel('Valor Real')
plt.title('Heatmap: Valores Reais vs. Valores Previstos')
plt.show()
```
